```python
!pip install squarify

import squarify

sizes = df.groupby('State')['Profit'].sum().sort_values(ascending=True)

labels = [f'{cat}\n${val/1e3:.1f}K' for cat, val in zip(sizes.index, sizes.values)]

colors = sns.color_palette("Greens", n_colors=len(sizes))

plt.figure(figsize=(12, 12))
squarify.plot(
    sizes=sizes.values,
    label=labels,
    color=colors,
    alpha=0.9
)
plt.axis('off')
plt.title('Most Profitable State', fontsize=14)
plt.show()
```

**Most Profitable States**

* New Mexico: $799.0K

* California: $31,785.0K

* Illinois: $30,821.0K

**Interpretation**

California leads with the highest profit, followed by Illinois and New York, showing that these states are the key contributors to overall profitability.

New Mexico has the lowest profit among the state, indicating either a smaller market size or lower demand for the product in that state.

```python
df.groupby('Product')['Profit Margin'].mean().nlargest(5).reset_index()
```

```python
profit_margin_per_product = df.groupby('Product')['Profit Margin'].mean().nlargest(5).reset_index()

plt.figure(figsize=(8, 5))
bars = plt.barh(profit_margin_per_product['Product'], profit_margin_per_product['Profit Margin'], color='Green')
plt.title('Top 5 Products with Highest Profit Margin')
plt.xlabel('Profit Margin')
plt.ylabel('')

ax = plt.gca()
for side in ['top', 'right']:
    ax.spines[side].set_visible(False)
ax.spines['left'].set_color('gray')
ax.spines['bottom'].set_color('gray')
ax.spines['left'].set_linewidth(0.8)
ax.spines['bottom'].set_linewidth(0.8)

for bar in bars:
    width = bar.get_width()
    plt.text(
        width + 0.01,  
        bar.get_y() + bar.get_height() / 2,  
        f'{round(width, 2)}%',  
        ha='left', va='center', fontsize=11, color='black'
    )

plt.grid(False)
plt.tight_layout()
plt.show()
```

**Top 5 Products with Highest Profit Margin**

* Decaf Espresso: 36.94%

* Darjeeling: 35.16%

* Columbian: 33.87%

**Interpretation**

Decaf Espresso has the highest profit margin, followed closely by Darjeeling and Columbian. These products are the most efficient at converting sales into profit.

```python
df.groupby('Product')['Profit Margin'].mean().nsmallest(5).reset_index()
```

```python
profit_margin_per_product = df.groupby('Product')['Profit Margin'].mean().nsmallest(5).reset_index()

plt.figure(figsize=(8, 5))
bars = plt.barh(profit_margin_per_product['Product'], profit_margin_per_product['Profit Margin'], color='Green')
plt.title('Top 5 Products with Lowest Profit Margin')
plt.xlabel('Profit Margin')
plt.ylabel('')

ax = plt.gca()
for side in ['top', 'right']:
    ax.spines[side].set_visible(False)
ax.spines['left'].set_color('gray')
ax.spines['bottom'].set_color('gray')
ax.spines['left'].set_linewidth(0.8)
ax.spines['bottom'].set_linewidth(0.8)

for bar in bars:
    width = bar.get_width()
    plt.text(
        width + 0.01,  
        bar.get_y() + bar.get_height() / 2,  
        f'{round(width, 2)}%',  
        ha='left', va='center', fontsize=11, color='black'
    )

plt.grid(False)
plt.tight_layout()
plt.show()
```

**Top 5 Products with Lowest Profit Margin**

* Green Tea: -127.50%

* Caffe Mocha: 4.65%

* Mint: 9.80%

**Interpretation**

Green Tea has an unusually negative profit margin, which suggests it might be underpriced or overproduced. Immediate action should be taken to either increase its price or reduce associated costs.

**Final Insight**

Based on the insights above:

- Decaf Espresso, Darjeeling, and Columbian are the top products with the highest margins they’re good at turning sales into profit.

- Green Tea has a negative profit margin and needs a pricing or cost review ASAP.

- California, Illinois, and New York are the big contributors to profit.

- Smaller markets aren't doing great, but there’s room for growth with targeted marketing.

- There's a strong link between sales and profit. The more sales, the more profit. But low-margin products like Green Tea could be pulling down the numbers.

- Major markets lead in sales, but small markets can still grow with the right strategies.

Recommendations:

1. Focus on High-Profit Products
Promote products like Decaf Espresso and Darjeeling to boost profit without adding extra costs.

2. Fix Low-Profit Products
Green Tea needs a price hike or cost cut. If it doesn’t improve, think about discontinuing it.

3. Market More in Major Areas
Keep focusing on Major Markets since they drive most of the sales. But also consider segmenting those markets for more targeted campaigns.

4. Grow Small Market Sales
Invest in smaller markets by running ads and promos specific to those regions, or partner with local retailers and influencers.

5. Drive Sales to Increase Profit
Since sales = profit, focus on increasing sales:

- Offer special deals or bundles for high-margin products.

- Train the sales team to close more deals.
