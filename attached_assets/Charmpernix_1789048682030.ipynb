```python
def kategori_produk(nama):
    nama = str(nama).lower()

    # Jersey dulu supaya tidak ketimpa "set"
    if 'jersey' in nama:
        return 'Jersey'

    elif 'celana' in nama:
        return 'Celana'

    elif any(x in nama for x in ['kurta','koko','gamis','sarung','sirwal','kemko']):
        return 'Setelan Muslim'

    elif any(x in nama for x in ['jaket','rompi','scuba','outer']):
        return 'Jaket / Outer'

    elif 'tas' in nama:
        return 'Tas'

    elif 'topi' in nama or 'cap' in nama:
        return 'Topi'

    elif any(x in nama for x in ['running','training','senam','lari']):
        return 'Training / Sportwear'

    elif any(x in nama for x in ['bel','manset','gembok','mask','sorban','bandana','lampu','pompa']):
        return 'Aksesoris'

    elif 'set' in nama or 'paket' in nama:
        return 'Paket / Set'

    else:
        return 'Lainnya'

df['Kategori'] = df['Produk'].apply(kategori_produk)
```

```python
import re

def rapikan_nama(nama):
    nama = str(nama).lower()

    # 1️⃣ Hapus kata promosi
    hapus_kata = [
        'cuci gudang', 'promo', 'grosir', 'murah',
        'original', 'premium', 'terbaik', 'diskon'
    ]

    for kata in hapus_kata:
        nama = nama.replace(kata, '')

    # 2️⃣ Hapus brand umum
    hapus_brand = ['cpx', 'the hamza', 'monkl', 'cpx sport', 'cpx cycling']

    for brand in hapus_brand:
        nama = nama.replace(brand, '')

    # 3️⃣ Hapus karakter aneh
    nama = re.sub(r'[^a-zA-Z0-9\s]', ' ', nama)

    # 4️⃣ Hilangkan spasi berlebih
    nama = re.sub(r'\s+', ' ', nama).strip()

    # 5️⃣ Ambil 6 kata pertama saja
    nama = ' '.join(nama.split()[:6])

    return nama.title()

df['Nama_Produk'] = df['Produk'].apply(rapikan_nama)
```

```python
import re

def simplify_name(text):
    text = text.lower()

    # hapus kata umum
    remove_words = [
        'baju','kaos','sport','olahraga','pria','wanita',
        'dewasa','remaja','dan','untuk','dengan'
    ]

    for word in remove_words:
        text = text.replace(word, '')

    # hapus karakter ganda
    text = re.sub(r'\s+', ' ', text).strip()

    # ambil maksimal 4 kata pertama
    text = ' '.join(text.split()[:4])

    return text.title()

df['Nama_Produk'] = df['Nama_Produk'].apply(simplify_name)
```

```python
import re

def clean_product_name(text):
    text = str(text).lower()

    # kata yang mau dibuang
    remove_words = [
        'order', 'orderan', 'custom', 'tambah', 'beli',
        'hanya', 'pcs', 'dari', 'dan', 'untuk',
        'temukan', 'rasakan', 'akhir', 'tahun',
        'tidak', 'dapat', 'memperoleh', 'informasi'
    ]

    for word in remove_words:
        text = text.replace(word, '')

    # hilangkan kata dobel (tas tas, jersey jersey, dll)
    words = text.split()
    cleaned_words = []
    for w in words:
        if w not in cleaned_words:
            cleaned_words.append(w)

    text = ' '.join(cleaned_words)

    # hapus karakter aneh
    text = re.sub(r'[^a-zA-Z0-9\s]', '', text)
    text = re.sub(r'\s+', ' ', text).strip()

    # ambil maksimal 4 kata
    text = ' '.join(text.split()[:4])

    return text.title()

df['Nama_Produk'] = df['Nama_Produk'].apply(clean_product_name)
```

```python
cols_numeric = [
    'Pengunjung Produk (Kunjungan)',
    'Halaman Produk Dilihat',
    'Pengunjung Melihat Tanpa Membeli',
    'Klik Pencarian',
    'Suka',
    'Tingkat Konversi Produk Dimasukkan ke Keranjang',
    'Tingkat Konversi (Pesanan yang Dibuat)',
    'Tingkat Konversi (Pesanan Siap Dikirim)'
]

df[cols_numeric] = df[cols_numeric].apply(pd.to_numeric, errors='coerce')
```
